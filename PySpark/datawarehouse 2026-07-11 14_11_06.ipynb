DATA WAREHOUSING

```python
%sql
CREATE DATABASE SalesDWH;
```

```python
%sql
-- Initial load
-- Stagging layer
CREATE TABLE SalesDWH.stg_sales
AS
SELECT * FROM sales_order;
```

```python
%sql
SELECT * FROM salesdwh.stg_sales;
```

```python
%sql
-- tranformation
CREATE VIEW salesDWH.transformed_sales 
AS
SELECT * FROM salesdwh.stg_sales WHERE Quantity IS NOT NULL 
```

```python
%sql
-- core layer
CREATE TABLE SalesDWH.core_sales
AS
SELECT * FROM salesDWH.transformed_sales;
```

```python
%sql
SELECT * FROM salesdwh.core_sales
```

```python
%sql
-- CDC [Change Date Capture] -- Incremental load
-- 1. Identify new records
-- 2. Load new records to staging layer
-- 3. Identify updated records
-- 4. Load updated records to staging layer
-- staging layer
CREATE OR REPLACE TABLE SalesDWH.stg_sales
AS
SELECT * FROM sales_order WHERE orderId > 1020;
```

```python
%sql
-- transformation
SELECT * FROM SalesDWH.transformed_sales;
```

```python
%sql
-- core layer
CREATE OR REPLACE TABLE SalesDWH.core_sales
AS
SELECT * FROM SalesDWH.transformed_sales
```
